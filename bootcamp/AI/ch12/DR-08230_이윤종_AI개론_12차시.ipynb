## I. 단답형 문제

[단답형 1]

훈련 데이터(train data)에 대해서는 정확도가 높게 나오지만, 검증 데이터(validation data)에 대해서는 정확도가 낮게 나오는 현상을 무엇이라고 하며, 이 현상이 발생하기 시작하는 시점(Epoch)을 학습 곡선(Loss curve)에서 어떻게 판단할 수 있습니까?

답:

1. 현상: 과적합
2. 판단 시점: 학습 곡선에서 훈련에서의 손실은 감소하는 반면 검증에서의 손실은 다시 증가하는 시점에서 과적합이 발생함

[단답형 2]

12차시 강의안에서 배운 튜닝 기법 중, 훈련(training) 시와 평가(evaluation) 시에 다르게 동작해야 해서 `model.train()`과 `model.eval()` 모드 전환이 반드시 필요한 기법 2가지를 쓰시오.

답:

1. 드롭아웃:  훈련시엔 과적합 방지를 위해 뉴런 일부를 무적위로 끄고, 평가 시엔 모든 뉴런을 사용한다.
2. 배치 정규화: 미니배치 통계로 정규화하며, 전체 데이터의 통계치를 업데이트. 평가 시엔 누적한 이동평균을 사용한다.

[코드 1]

`CNN_v2` 모델의 최적화 함수(optimizer)를 `optim.SGD`에서 `optim.Adam`으로 변경하는 코드를 작성하십시오. (learning rate는 0.01로 동일하게 설정)

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.datasets as datasets
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
import numpy as np
```

```python
# 12차시 코드 셀의 optim, net 변수가 이미 선언되어 있다고 가정합니다.
# net = CNN_v2(n_output).to(device)
# lr = 0.01

class CNN_v2(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        # 합성곱, ReLU, MaxPool 레이어 정의
        self.conv1 = nn.Conv2d(3, 32, 3, padding=(1,1))
        # padding=(1,1) 이미지 크기 유지
        self.conv2 = nn.Conv2d(32, 32, 3, padding=(1,1))
        self.conv3 = nn.Conv2d(32, 64, 3, padding=(1,1))
        self.conv4 = nn.Conv2d(64, 64, 3, padding=(1,1))
        self.conv5 = nn.Conv2d(64, 128, 3, padding=(1,1))
        self.conv6 = nn.Conv2d(128, 128, 3, padding=(1,1))
        self.relu = nn.ReLU(inplace=True)
        self.flatten = nn.Flatten()
        self.maxpool = nn.MaxPool2d((2,2))

        # 선형(Fully Connected) 레이어 정의
        self.l1 = nn.Linear(4*4*128, 128)
        self.l2 = nn.Linear(128, num_classes)

        # 레이어들을 순차적으로 실행할 `nn.Sequential` 정의
        self.features = nn.Sequential(
            self.conv1,
            self.relu,
            self.conv2,
            self.relu,
            self.maxpool,
            self.conv3,
            self.relu,
            self.conv4,
            self.relu,
            self.maxpool,
            self.conv5,
            self.relu,
            self.conv6,
            self.relu,
            self.maxpool,
            )

        self.classifier = nn.Sequential(
            self.l1,
            self.relu,
            self.l2
        )

    def forward(self, x):
        x1 = self.features(x)
        x2 = self.flatten(x1)
        x3 = self.classifier(x2)
        return x3
    
lr = 0.001

classes = ('plane', 'car', 'bird', 'cat',
           'deer', 'dog', 'frog', 'horse', 'ship', 'truck')

n_output = len(list(set(classes)))

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

net = CNN_v2(n_output).to(device)

# TODO: 최적화 함수를 optim.Adam으로 변경하시오.
optimizer = optim.Adam(net.parameters(), lr=lr)
```

`CNN_v2` 모델의 분류기(classifier) 부분에서, 첫 번째 선형 레이어(`nn.Linear(2048, 128)`)와 활성화 함수(`self.relu`) *다음에* 25%의 비율로 `nn.Dropout` 레이어를 추가하여 `CNN_v3` 모델을 완성하십시오.

```python
# 12차시 코드 셀의 nn, CNN_v2, self.relu 등이 이미 선언되어 있다고 가정합니다.

class CNN_v3(nn.Module):
    def __init__(self, n_output):
        super().__init__()
        # ... (features 부분은 CNN_v2와 동일하다고 가정) ...
        # self.features = ...
        # self.relu = nn.ReLU(inplace=True)
        self.features = nn.Sequential(
            self.conv1,
            self.relu,
            self.conv2,
            self.relu,
            self.maxpool,
            self.conv3,
            self.relu,
            self.conv4,
            self.relu,
            self.maxpool,
            self.conv5,
            self.relu,
            self.conv6,
            self.relu,
            self.maxpool,
            )

        self.classifier = nn.Sequential(
            nn.Linear(2048, 128),
            self.relu,
            # TODO: 여기에 25% 비율의 드롭아웃 레이어를 추가하시오.
            nn.Dropout(p=0.25),
            nn.Linear(128, n_output)
        )

    def forward(self, x):
        # ... (forward 함수는 CNN_v2와 동일하다고 가정) ...
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x
```
