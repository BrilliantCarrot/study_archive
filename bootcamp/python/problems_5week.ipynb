```python
# 문제 1) 아래 그래프에서 DFS로 그래프를 읽을 시 6번째로 출력되는 데이터를 쓰시오.

myGraph = {
    '두': ['산', '로', '키'],
    '산': ['두', '부'],
    '로': ['두'],
    '키': ['두', '트'],
    '부': ['산'],
    '트': ['키', '캠', '프'],
    '캠': ['트'],
    '프': ['트']
}

def my_dfs(graph, start_node):
    stack = list()
    visited = list()

    stack.append(start_node)

    while stack:
        node = stack.pop()

        if node not in visited:
            stack.extend(reversed(graph[node]))
            visited.append(node)

    print("dfs - ", visited)
    return visited

my_dfs(myGraph, '두')
```

```python
# 문제 2) 아래 그래프에서 BFS로 그래프를 읽을 시 6번째로 출력되는 데이터를 쓰시오.

myGraph = {
    '두': ['산', '로', '키'],
    '산': ['두', '부'],
    '로': ['두'],
    '키': ['두', '트'],
    '부': ['산'],
    '트': ['키', '캠', '프'],
    '캠': ['트'],
    '프': ['트']
}

def my_bfs(graph, start_node):
    queue = list()
    visited = list()
    queue.append(start_node)
    while queue:
        node = queue.pop(0)
        if node not in visited:
            queue.extend(graph[node])
            visited.append(node)

    print("bfs - ", visited)
    return visited

my_bfs(myGraph, '두')
```

```python
# 문제 1) 다음은 공공데이터 포털에서 제공하는 대구광역시 동구_연도별 체납차량 번호판 영치실적 데이터 링크이다. 
# 해당 링크에서 대구광역시 동구_연도별 체납차량 번호판 영치실적 데이터 csv 파일을 다운받고 
# 해당 csv 파일을 pandas를 이용해 데이터 프레임으로 변경하는 프로그램을 작성하시오.
# https://www.data.go.kr/data/15094034/fileData.do

import pandas as pd

df = pd.read_csv('대구광역시 동구_연도별 체납차량 번호판 영치실적_20251031.csv',
                 encoding='euc-kr')

print(df)
print(df.dtypes)
```

```python
# 위에서 생성한 데이터 프레임을 활용하여 2016년 부터 2024년 까지 연도별 영치대수 선 그래프를 matplotlib을 이용해 그리십시오.

import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

df_filtered = df[(df['년도'] >= 2016) & (df['년도'] <= 2024)]

plt.plot(df_filtered['년도'], df_filtered['영치대수'])
for x, y in zip(df_filtered['년도'], df_filtered['영치대수']):
    plt.text(x, y + 20, str(y), ha='center', fontsize=9)
plt.title('2016 ~ 2024 연도별 영치대수')
plt.xlabel('년도')
plt.ylabel('영치대수')
plt.xticks(df_filtered['년도'])
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
