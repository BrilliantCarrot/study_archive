```python
# 클래스를 통한 스택 구현

class Stack:
    def __init__(self):
        self.stack = []

    def push(self, data):
        self.stack.append(data)

    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        return
    
    def is_empty(self):
        if len(self.stack) == 0:
            return True
        return False
    
    def peak(self):
        if not self.is_empty():
            print("스택 맨 위에 있는 값:", end='')
            return self.stack[-1]
        return
    
    def status_check(self):
        return self.stack
    
s1 = Stack()
print(s1.peak())
s1.pop
s1.push(1)
s1.push(2)
s1.pop()
s1.push(3)
s1.push(4)
print(s1.peak())
print(s1.status_check())
```

```python
# 클래스를 통한 큐 구현

class Queue:
    def __init__(self):
        self.queue = []

    def enqueue(self, data):
        self.queue.append(data)

    def dequeue(self):
        if not self.is_empty():
            return self.queue.pop(0)
        return
    
    def is_empty(self):
        if len(self.queue) == 0:
            return True
        return False
    
    def status_queue(self):
        return self.queue
    
q1 = Queue()
q1.dequeue()
q1.enqueue(1)
q1.enqueue(2)
q1.dequeue()
print(q1.status_queue())
q1.enqueue(3)
q1.enqueue(4)
q1.dequeue()
print(q1.status_queue())
```

```python
# 문제1: 괄호 짝 검사
# 문자열을 인수로 받아 포함된 괄호의 짝이 올바르게 사용되었는지 확인하는 프로그램을 작성하시오
# 스택 활용, 괄호의 짝이 맞으면 True 반환, 괄호의 짝이 맞지 않으면 False 반환

def check_parentheses(text):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}

    for ch in text:
        if ch in '([{':
            stack.append(ch)
        elif ch in ')]}':
            if not stack:
                return False
            if stack[-1] != pairs[ch]: # 지금 닫는 괄호가, 스택 맨 위 여는 괄호와 짝이 맞는지 검사, 즉
                # pairs[')'] → '('
                # pairs[']'] → '['
                # pairs[']'] → '['
                return False
            stack.pop()

    return len(stack) == 0

print(check_parentheses("()"))          # True
print(check_parentheses("(())"))        # True
print(check_parentheses("([{}])"))      # True
print(check_parentheses("(()"))         # False
print(check_parentheses("([)]"))        # False
print(check_parentheses("())"))         # False
```

```python
# 문제2: 회전 큐 구현
# 데이터를 양방향으로 삽입하거나 제거할 수 있는 회전 큐를 구현하시오
# Collections.deque로 구현
from collections import deque

class RotatingQueue:
    def __init__(self):
        self.queue = deque()

    def push_front(self, value):
        self.queue.appendleft(value)

    def push_back(self, value):
        self.queue.append(value)

    def pop_front(self):
        if self.is_empty():
            return None
        return self.queue.popleft()
    
    def pop_back(self):
        if self.is_empty():
            return None
        return self.queue.pop()
    
    def rotate_left(self, n=1):
        self.queue.rotate(-1) # 양수면 오른쪽 회전, 음수면 왼쪽 회전

    def rotate_right(self, n=1):
        self.queue.rotate(n)

    def is_empty(self):
        return len(self.queue) == 0
    
    def __str__(self):
        return str(list(self.queue))
    

# 리스트로 직접 구현
class RotatingQueue:
    def __init__(self):
        self.queue = []

    def push_front(self, value):
        self.queue.insert(0, value)

    def push_back(self, value):
        self.queue.append(value)

    def pop_front(self):
        if not self.queue:
            return None
        return self.queue.pop(0)

    def pop_back(self):
        if not self.queue:
            return None
        return self.queue.pop()

    def rotate_left(self, n=1):
        for _ in range(n):
            if self.queue:
                first = self.queue.pop(0)
                self.queue.append(first)

    def rotate_right(self, n=1):
        for _ in range(n):
            if self.queue:
                last = self.queue.pop()
                self.queue.insert(0, last)

    def __str__(self):
        return str(self.queue)

# 테스트
rq = RotatingQueue()

rq.push_back(1)
rq.push_back(2)
rq.push_back(3)
print(rq)   # [1, 2, 3]

rq.push_front(0)
print(rq)   # [0, 1, 2, 3]

rq.rotate_right(1)
print(rq)   # [3, 0, 1, 2]

rq.rotate_left(2)
print(rq)   # [1, 2, 3, 0]

print(rq.pop_front())   # 1
print(rq.pop_back())    # 0
print(rq)               # [2, 3]
```

```python
# 7. 후위 표기법(Postfix Notation, Reverse Polish Notation)으로 주어진 수식을 계산하는 프로그램을 작성하세요.
# 후위 표기법에서는 연산자가 피연산자 뒤에 위치합니다. 예를 들어, 3 4 +는 3 + 4를 의미하며, 결과는 7입니다.
# 연산자는 +, -, *, /만 고려합니다. (복수 연산자도 처리 가능해야 함.)

def postfix_calc(expression):
    stack = []
    tokens = expression.split()  # 공백 기준으로 토큰 분리

    for token in tokens:
        if token not in ('+', '-', '*', '/'):
            # 숫자면 스택에 push (float으로 변환해 소수점도 처리)
            stack.append(float(token))
        else:
            # 연산자면 스택에서 피연산자 2개 pop
            # 주의: 먼저 꺼낸 게 오른쪽 피연산자
            b = stack.pop()   # 두 번째 피연산자 (오른쪽)
            a = stack.pop()   # 첫 번째 피연산자 (왼쪽)

            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                stack.append(a / b)

    return stack[0]  # 최종 결과는 스택에 하나 남음

print(postfix_calc("3 4 +"))
print(postfix_calc("5 1 2 + 4 * +"))
print(postfix_calc("3 4 + 2 *"))
print(postfix_calc("10 2 /"))
```

```python
# 8. 리스트를 사용하여 기본적인 큐(queue)를 구현하세요. (다음 내용을 고려할 것.)
# enqueue(x): 정수 x를 큐의 끝에 추가
# dequeue(): 큐의 앞에서 값을 제거하고 반환. 만약 큐가 비어 있다면 -1을 반환
# front(): 큐의 앞에 있는 값을 반환. 만약 큐가 비어 있다면 -1을 반환
# is_empty(): 큐가 비어 있으면 True, 아니면 False를 반환

queue = []
def enqueue(x):
    queue.append(x)

def dequeue():
    if is_empty():
        return -1
    return queue.pop(0)

def front():
    if is_empty():
        return -1
    return queue[0]

def is_empty():
    if len(queue) == 0:
        return True
    else:
        return False
    
enqueue(1); enqueue(2); enqueue(3)
print(front())    # 1
print(dequeue())  # 1
print(queue)
print(dequeue())  # 2
print(dequeue())  # 3
print(dequeue())  # -1 (빈 큐)
```
