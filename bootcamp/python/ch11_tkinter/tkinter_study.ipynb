```python
# 엔트리에 값 입력하고 확인 버튼을 누르면 라벨에 업데이트 

otk = Tk()
otk.geometry("400x100+3000+400")

ostring = StringVar()
display = StringVar()

def on_click():
    display.set(ostring.get())

oentry = Entry(otk, textvariable=ostring)
olabel = Label(otk, textvariable=display)
obtn1 = Label(otk, text = "적색", bg = "red", width= 20)
obtn2 = Label(otk, text = "녹색", bg = "green", width=30)
obtn3 = Label(otk, text = "파란색", bg = "blue", width=40)
obtn4 = Button(otk, text = "확인", command=on_click)

oentry.pack()
olabel.pack()
obtn1.pack()
obtn2.pack()
obtn3.pack()
obtn4.place(x=10, y=30)

otk.mainloop()
```

```python
oroot = Tk()



oroot.geometry("400x400")

radio_value = IntVar() # 정수형 변수 생성
radio_value.set(-1) # 정수값 설정, -1은 아무것도 설정 안됨으로 설정

lunch = {0:"A런치", 1:"B런치", 2:"C런치"}

def buy():
    rValue = radio_value.get()
    print(f"{lunch[rValue]}를 구입하였습니다")


# variable: 클릭된 버튼의 정보를 저장할 변수명 설정
orbtn1 = Radiobutton(oroot, text=lunch[0], variable=radio_value, value=0)
orbtn2 = Radiobutton(oroot, text=lunch[1], variable=radio_value, value=1)
orbtn3 = Radiobutton(oroot, text=lunch[2], variable=radio_value, value=2)
orbtn1.pack()
orbtn2.pack()
orbtn3.pack()



    
obtn = Button(oroot, text="주문", command = buy)
obtn.pack()

oroot.mainloop()
```

```python
from tkinter import *

window = Tk()
window.title("조각 피자 주문 프로그램")
window.geometry("400x350")

Label(window, text="피자").pack()

cheese_var = IntVar()
combo_var = IntVar()
bulgogi_var = IntVar()

Checkbutton(window, text="치즈 피자 (3200원)", variable=cheese_var).pack(anchor="w")
Checkbutton(window, text="콤비네이션 피자 (3500원)", variable=combo_var).pack(anchor="w")
Checkbutton(window, text="불고기 피자 (3600원)", variable=bulgogi_var).pack(anchor="w")

def order():
    total = 0
    result_text = "주문내역:\n"

    if cheese_var.get() == 1:
        result_text += "- 치즈 피자 (3200원)\n"
        total += 3200

    if combo_var.get() == 1:
        result_text += "- 콤비네이션 피자 (3500원)\n"
        total += 3500

    if bulgogi_var.get() == 1:
        result_text += "- 불고기 피자 (3600원)\n"
        total += 3600

    result_text += f"\n총 가격: {total}원"

    result_label.config(text=result_text)

Button(window, text="주문", command=order).pack()
result_label = Label(window, text="")
result_label.pack()

window.mainloop()
```

1. tkinter의 위젯 배치 메서드는 pack(), grid(), place() 세 가지  
1.1 pack()은 위젯을 위, 아래, 왼쪽, 오른쪽 방향으로 차례대로 배치하는 방식이라서 가장 단순한 화면 구성에 잘 맞습니다.
버튼 몇 개를 세로로 쌓거나, 좌우로 나누는 정도의 간단한 UI에 쓰기 좋지만 복잡한 입력 폼을 만들기엔 불편합니다.  
1.2 grid()는 위젯을 행(row)과 열(column) 기준으로 배치하므로, 엑셀 표 같은 구조나 로그인 창, 입력 폼에 적합합니다.
예를 들어 “이름-입력칸”, “비밀번호-입력칸”처럼 정렬이 중요한 화면은 grid()가 가장 깔끔합니다.  
1.3 place()는 x, y 좌표를 직접 지정해서 원하는 위치에 놓는 방식이라 가장 자유도가 높습니다.
다만 창 크기가 바뀌면 배치가 어색해지거나 깨지기 쉬워서, 일반적인 폼 화면보다는 위치를 아주 정확히 잡아야 하는 경우에 더 적합합니다.

2. Button 위젯에서 command 매개변수를 사용하여 버튼 클릭 시 실행할 함수를 지정

3. Radiobutton의 특징으로 옳은 것: 하나만 선택 가능한 위젯, 여러개 선택하려면 Checkbotton을 사용

```python
# 문제8: Radiobutton 활용
# 1. IntVar() 변수 생성 및 초기값 0 설정
# 2. "Python", "Java", "C++" 텍스트의 Radiobutton 3개 생성
# 3. 각 Radiobutton의 value를 0, 1, 2로 설정
# 4. "선택" 버튼 생성
# 5. 버튼 클릭 시 선택된 언어를 출력하는 함수 정의
# 6. pack()으로 배치

from tkinter import *

def show_selection():
    languages = {0:"Python", 1:"Java", 2:"C++"}
    selected = radio_var.get()
    print(languages[selected])

window = Tk()
window.geometry("400x200")

radio_var = IntVar()
radio_var.set(0)

button1 = Radiobutton(window, text="Python", variable=radio_var, value=0)
button2 = Radiobutton(window, text="Java", variable=radio_var, value=1)
button3 = Radiobutton(window, text="C++", variable=radio_var, value=2)
button1.pack()
button2.pack()
button3.pack()
button = Button(window, text="선택", command=show_selection)
button.pack()
window.mainloop()
```
