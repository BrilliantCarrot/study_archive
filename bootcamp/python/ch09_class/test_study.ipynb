1. 객체지향 프로그래밍과 클래스 설명
객체지향 프로그래밍(OOP)은 프로그램을 데이터와 기능을 함께 가진 “객체” 중심으로 구성하는 방식입니다.

이 방식은 코드의 재사용성, 가독성, 유지보수성을 높이는 데 도움이 되며, 클래스는 이러한 객체를 만들기 위한 설계도 역할을 합니다.

클래스(class)는 속성(attribute, 멤버 변수)과 메서드(method, 멤버 함수)를 정의하는 틀이며, 객체(object)는 그 클래스로부터 생성된 실제 데이터입니다.

예를 들어 Dog라는 클래스를 만들고 이름과 나이를 속성으로 정의하면, Dog("Bori", 3) 같은 객체를 생성해 사용할 수 있습니다.

2. 클래스 구성 요소 및 문법 설명
파이썬 클래스는 class 클래스이름: 형식으로 정의하며, 내부에는 속성과 메서드를 작성합니다.

속성은 객체가 가지는 데이터이고, 메서드는 객체가 수행하는 동작입니다.

기본 문법은 아래와 같습니다.

python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"이름은 {self.name}이고, 나이는 {self.age}살입니다.")
위 코드에서 Person은 클래스 이름이고, name, age는 객체가 가지는 속성이며, introduce()는 객체의 정보를 출력하는 메서드입니다.

3. 객체 생성 및 사용 코드 작성
객체는 클래스이름() 형식으로 생성합니다. 객체를 만들면 클래스에 정의된 속성과 메서드를 사용할 수 있습니다.

python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"이름: {self.name}, 나이: {self.age}")

p1 = Person("홍길동", 25)
p1.introduce()
위 코드에서 p1은 Person 클래스의 객체이며, introduce() 메서드를 호출해 객체의 정보를 사용할 수 있습니다.

4. self 키워드 설명 및 사용 방법
self는 현재 생성되었거나 사용 중인 인스턴스 객체 자신을 가리키는 참조입니다.

파이썬 클래스의 인스턴스 메서드에서는 첫 번째 매개변수로 사용하며, 이를 통해 객체의 속성이나 다른 메서드에 접근합니다.

​

self라는 이름은 관례적으로 사용하는 것이며, technically 다른 이름도 가능하지만 일반적으로 self를 사용합니다.

5. __init__() 메서드를 사용한 객체 초기화
__init__() 메서드는 객체가 생성될 때 자동으로 호출되는 초기화 메서드입니다.

주로 객체가 처음 만들어질 때 필요한 속성 값을 저장하거나 초기 상태를 설정하는 데 사용합니다.

예를 들어 아래 코드는 학생 객체가 생성될 때 이름과 점수를 바로 저장합니다.

python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def show_info(self):
        print(f"{self.name}의 점수는 {self.score}점입니다.")

s1 = Student("김철수", 90)
s1.show_info()
이처럼 __init__()를 사용하면 객체를 만들 때 필요한 값을 한 번에 전달하여 초기화할 수 있습니다.

6. 내장 클래스와 사용자 정의 클래스 구분
내장 클래스는 파이썬이 기본으로 제공하는 클래스이며, 대표적으로 int, str, list, dict 등이 있습니다.

사용자 정의 클래스는 프로그래머가 문제 해결을 위해 직접 class 문으로 만든 클래스입니다.

예를 들어 a = 10에서 10은 int 클래스의 객체이고, "python"은 str 클래스의 객체입니다.

반면 아래 Animal 클래스는 사용자가 직접 정의한 클래스입니다.

python
class Animal:
    def __init__(self, name):
        self.name = name

dog = Animal("초코")
즉, 내장 클래스는 이미 제공되는 것, 사용자 정의 클래스는 직접 만드는 것이라고 구분하면 됩니다.

#### self  
해당 클래스의 객체를 의미  
클래스 내 메서드의 첫번째 매개변수로 사용  
클래스를 통해 생성된 객체를 구분하기 위해 사용  
self 객체를 통해 변수와 메서드를 호출 가능  

```python
class Cadd:
    def fadd(self, a, b):
        self.x = a
        self.y = b
        self.hap = self.x + self.y

obj = Cadd()
obj.fadd(10,20)
print("객체 obj 내의 인스턴스 변수 x의 값은: ", obj.x)
print("객체 obj 내의 인스턴스 변수 x의 값은: ", obj.y)
print("객체 obj 내의 인스턴스 변수 x의 값은: ", obj.hap)
```

```python
# 코딩 문제 2

class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.scores = []

    def add_score(self, score):
        self.scores.append(score)
    
    def get_average(self):
        total = 0
        length = len(self.scores)
        for i in range(len(self.scores)):
            total = total + self.scores[i]
        avg = total/length
        return avg
    
student = Student("김철수", 2024001)
student.add_score(85)
student.add_score(90)
student.add_score(88)
print(f"평균 : {student.get_average()}")
```

```python
# 코딩 문제 3

class Calculator:
    def __init__(self):
        self.result = 0
    def add(self, num):
        self.result += num
    def subtract(self, num):
        self.result -= num
    def multiply(self, num):
        self.result = self.result * num
    def reset(self):
        self.result = 0
    def show_result(self):
        print(f"결과: {self.result}")

calc = Calculator()
calc.add(10)
calc.multiply(5)
calc.subtract(20)
calc.show_result()
```
