#### 예외 처리하기  
try ... except 구문   
try은 다른 예외의 처리를 위한 여러 개의 except 절을 가질 수 있음   
최대 하나의 핸들러만 실행   
except 절은 여러 예외를 괄호로 묶은 튜플로 명명 가능  
예) except (RuntimeError, TypeError, NameError):  
각 예외에는 예외 관련  구체적 정보가 저장됨   
예외 정보는 except 절의 예외 이름 뒤 변수를 as로 지정하여 확인 가능   
except Exception as e:    
&nbsp;&nbsp;&nbsp;&nbsp;  print(e)   

모든 예외의 공통 기본 클래스: BaseException
실행 중 감지되는 예외의 기본 클래스: Exception

람다 함수 & map() 함수  
람다 함수: 이름이 지정되지 않은 함수, def 키워드를 사용하지 않고도 함수를 정의 가능  
map()함수: iteralbe의 각 요소에 함수를 적용하여 결과를 반환한다  
map()함수의 결과를 리스트로 반환하려면 list()로 감싸서 변환해야 한다.

```python
f = lambda x, y : x+y
print(f(3,5))
```
