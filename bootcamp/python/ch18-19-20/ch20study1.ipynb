```python
# import os
# os.getcwd()

import cv2

image = cv2.imread('Ocean_18_OBGA.SumilonIslandPhilippines_AdobeStock_152111654_Ocean.jpg')
image = cv2.resize(image, (1200, 600))
cv2.imshow("Loaded Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image',gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 이미지 처리 기초 

# 이미지 회전 
# cv2.getRotationMatrix2D(center, angle, scale) -> retval 
# » 회전 변환 행렬을 생성 (임의의 행렬을 원점을 중심으로 회전) 
# center: 회전 중심 좌표. (x, y) 튜플 
# angle: (반시계 방향) 회전 각도(degree). 음수는 시계 방향 
# scale: 추가적인 확대 비율 
# retval: 2x3 어파인 변환 행렬. 실수형 

# warpAffine(src, M, dsize) -> dst 
# » 이미지를 변환 
# src: 입력 영상 
# M: 2x3 어파인 변환 행렬. 실수형 
# dsize: 결과 영상 크기. (w, h) 튜플. (0, 0)이면 src와 같은 크기로 설정 
# dst: 출력 영상

(h, w) = image.shape[:2] 
center = (w // 2, h // 2) 
M = cv2.getRotationMatrix2D(center, 45, 1.0) 
rotated = cv2.warpAffine(image, M, (w, h)) 
cv2.imshow('Rotated Image', rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 8. OpenCV를 사용하여 이미지를 불러오고 화면에 출력하세요.
# cv2.imread()로 이미지를 로드한 후, 
# cv2.imshow()를 사용하여 출력하세요.
# cv2.waitKey(0)을 사용하여 창을 닫을 수 있도록 하세요.

import cv2

image = cv2.imread('Ocean_18_OBGA.SumilonIslandPhilippines_AdobeStock_152111654_Ocean.jpg')
image = cv2.resize(image, (1200, 600))
cv2.imshow("Loaded Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 9. OpenCV를 사용하여 컬러 이미지를 흑백(그레이스케일)으로 변환하세요.
# cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)를 사용하여 변환하세요.
# 변환된 이미지를 출력하세요.
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image',gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 10. OpenCV의 cv2.Canny()를 사용하여 이미지의 엣지를 감지하세요.
# 엣지를 감지한 후, 결과 이미지를 출력하세요.

edges = cv2.Canny(gray, threshold1=100, threshold2=200)
cv2.imshow('Edge Detection', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
