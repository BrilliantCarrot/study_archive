#### 정규표현식  
주로 텍스트 검색, 매칭, 추출, 데이터 변환에 사용    
메타 문자: 원래 해당 문자가 가진 뜻이 아닌 특별한 의미를 가진 문자  

① 메타문자 (특수 의미를 가진 문자)  
.	임의의 문자 1개 (줄바꿈 제외), 줄바꿈 문자인 \n을 제외한 모든 문자와 매치된다는것을 의미    
a.b면 a와 b 사이 어떤 문자가 들어가도 모두 매치 됨  
^	문자열의 시작   
$	문자열의 끝     
( * )	앞 문자가 0회 이상 반복     
( + )	앞 문자가 1회 이상 반복     
?	앞 문자가 0회 또는 1회      
{n}	앞 문자가 정확히 n회 반복   
{n,m}	앞 문자가 n회 이상 m회 이하 
[]	문자 클래스 — 괄호 안 문자 중 하나  
[^...]	[] 안에서 ^ : 부정 (해당 문자 제외)     
|	OR (또는)   
()	그룹 지정   
\	이스케이프 (메타문자를 일반 문자로) 

② 자주 쓰는 축약 표현 (Shorthand)   
표현	의미    
\d	숫자 [0-9]      
\D	숫자가 아닌 문자    
\w	단어 문자 숫자 [a-zA-Z0-9_]  
\W	단어 문자 숫자가 아닌 것     
\s	공백 문자 (스페이스, 탭, 줄바꿈)        
\S	공백이 아닌 문자    
\b	단어 경계 (Word Boundary)   

③ re 모듈 주요 함수 
함수	설명    
re.match(p, s)	문자열 s의 처음부터 패턴 p 매칭. 성공 시 match 객체 반환    
re.search(p, s)	문자열 전체에서 패턴 p 검색. 처음 찾은 위치 반환    
re.findall(p, s)	패턴 p와 일치하는 모든 문자열을 리스트로 반환   
re.finditer(p, s)	패턴 p와 일치하는 모든 match 객체를 iterator로 반환     
re.sub(p, r, s)	패턴 p와 일치하는 부분을 r로 교체   
re.split(p, s)	패턴 p를 구분자로 문자열 s를 분리   
re.compile(p)	패턴 p를 컴파일. 반복 사용 시 성능 향상     

④ match 객체 메서드     
메서드	설명    
.group()	매칭된 문자열 반환  
.start()	매칭 시작 인덱스    
.end()	매칭 끝 인덱스  
.span()	(start, end) 튜플 반환  

⑤ 컴파일 옵션 (flags)   
옵션	설명    
re.IGNORECASE (re.I)	대소문자 무시   
re.MULTILINE (re.M)	^, $ 가 각 줄의 시작/끝에 적용  
re.DOTALL (re.S)	'.'이 줄바꿈 문자도 포함    
re.VERBOSE (re.X)	패턴에 공백, 주석 허용 (가독성)     

⑥ match vs search 핵심 차이     
import re   
s = 'hello world'   

re.match(r'world', s)   # → None  (처음부터 매칭 안 됨)     
re.search(r'world', s)  # → match  (중간에서 찾음)  

⑦ 역슬래시 주의 — raw string 사용 권장  
❌ 일반 문자열: \d 가 파이썬 이스케이프로 처리될 수 있음  
re.findall('\d+', text)     

✅ raw string: r'' 사용 → \d가 정규식 의미 그대로 전달    
re.findall(r'\d+', text)    

match 객체는 어디서 찾았는지 정보를 담고있는 상자라고 생각

```python
# 문제 1 다음 텍스트에서 전화번호를 –(하이픈)을 제외한 숫자만 추출하시오. 
text = "My phone number is 123-456-7890"
numbers = re.findall(r"\d+", text) 
print(numbers)
```

```python
# 다음 텍스트에서 이메일 주소를 추출하시오. 
text = "Contact us at support@example.com or sales@example.org." 
pattern = r"[\w.-]+@[\w.-]+\.\w+" 
emails = re.findall(pattern, text) 
print(emails)
```

```python
#  다음 휴대폰 전화번호가 적절하게 사용되었는지 검증하시오. 
phone = "123-456-7890" 
phone_pattern = r"^\d{3}-\d{3}-\d{4}$" 
if re.match(phone_pattern, phone): 
   print("Valid phone number") 
else: 
   print("Invalid phone number") 
```

```python
# Q9.  match 객체의 .group(), .start(), .end() 메서드를 모두 사용하는 코드를 작성하세요.
text = 'Hello Python World'

import re
text = 'Hello Python World'
m = re.search(r'Python', text)
if m:
    print('매칭 문자열:', m.group())    # Python
    print('시작 인덱스:', m.start())    # 6
    print('끝 인덱스 :', m.end())      # 12
    print('범위       :', m.span())     # (6, 12)

```
