```python
total = 0
def add_to_total(num):
    global total
    total = total + num

add_to_total(10)
print(total)
add_to_total(20)
print(total)
add_to_total(30)
print(total)
```

```python
def countdown(n):
    if n == 0:
        print("발사!")
    else:
        print(n)
        countdown(n-1)

countdown(5)
```
